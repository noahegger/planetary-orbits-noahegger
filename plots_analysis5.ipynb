```python
import pylab
import numpy as np 
import matplotlib.pyplot as plt
import f90nml

plt.style.use('dark_background')
```

```python
parameters_1 = f90nml.read('circular.namelist')
file_name_1 = parameters_1['output']['output_file']
all_data = np.loadtxt(file_name_1,skiprows=1)
all_data = all_data.transpose()

time = all_data[0]
x_1 = all_data[1]
y_1 = all_data[2]
x_2 = all_data[3]
y_2 = all_data[4]
energy = all_data[5]
momentum = all_data[6]

```

## Energy as a Function of Time for Two-Planet System in Circular Orbit

```python
plt.plot(time, energy)
pylab.ylim([energy[1]-0.001*energy[1],energy[1]+0.001*energy[1]])
plt.title('Energy for 2-Planet System')
plt.xlabel('Time ')
plt.ylabel('Total Energy')
plt.show()
```

## Angular Momentum as a Function of Time for Two-Planet System in Circular Orbit

```python
plt.plot(time, momentum, color = 'c')

pylab.ylim([momentum[1]-0.001*momentum[1],momentum[1]+0.001*momentum[1]])
# Limit on y-value is imposed as +/- 1.0% of the value of the momentum, in order
# to more easily see its conservation. 
plt.title('Angular Momentum for 2-Planet System')
plt.xlabel('Time ')
plt.ylabel('Total Ang. Momentum')
#plt.legend(loc='upper right')
plt.show()
```

## Two-Planet System Circular Orbit Plot

```python
plt.plot(x_1, y_1, Label='Planet One', color = 'c')
plt.plot(x_2, y_2, Label='Planet Two', color = 'm')
plt.plot(0,0,marker='o', Label='Star', color = 'y')
plt.title('Two-Planet System Orbits')
plt.xlabel('x position ')
plt.ylabel('y position')

plt.legend(loc='upper right')
plt.show()
```

```python
parameters_2 = f90nml.read('planet_moon.namelist')
file_name_2 = parameters_2['output']['output_file']
all_data = np.loadtxt(file_name_2,skiprows=1)
all_data = all_data.transpose()

time_pm = all_data[0]
x_1_pm = all_data[1]
y_1_pm = all_data[2]
x_2_pm = all_data[3]
y_2_pm = all_data[4]
energy_pm = all_data[5]
momentum_pm = all_data[6]
```

## Energy as a Function of Time for Planet-Moon System

```python
plt.plot(time_pm, energy_pm, color = 'c')
pylab.ylim([energy_pm[1]-0.0005*energy_pm[1],energy_pm[1]+0.0005*energy_pm[1]])
plt.title('Energy for Planet-Moon System')
plt.xlabel('Time ')
plt.ylabel('Total Energy')
plt.show()
```

## Angular Momentum as a Function of Time for Planet-Moon System

```python
plt.plot(time_pm, momentum_pm, color = 'c')

pylab.ylim([momentum_pm[1]-0.001*momentum_pm[1],momentum_pm[1]+0.001*momentum_pm[1]])
plt.title('Angular Momentum for Planet-Moon System')
plt.xlabel('Time ')
plt.ylabel('Total Ang. Momentum')
plt.show()
```

## Planet-Moon System Orbit Plot

```python
plt.plot(x_1_pm, y_1_pm, Label='Planet', color = 'w')
plt.plot(x_2_pm, y_2_pm, Label='Moon', color = 'c')
plt.plot(0,0,marker='o', Label='Star', color = 'y')

plt.title('Planet-Moon System Orbits')
plt.xlabel('x position ')
plt.ylabel('y position')

plt.legend(loc='upper right')
plt.show()
```

## Close-up Planet-Moon System Orbit Plot

```python
plt.plot(x_1_pm, y_1_pm, marker='o', Label='Planet', color = 'w')
plt.plot(x_2_pm, y_2_pm, Label='Moon', color = 'c')

pylab.ylim(175,210)
pylab.xlim(-50,50)

plt.title('Close-Up Planet-Moon System')
plt.xlabel('x position ')
plt.ylabel('y position')

plt.legend(loc='upper right')
plt.show()
```
